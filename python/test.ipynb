```python
origins = sample['origin'].drop_duplicates().reset_index(drop=True)
```

```python
origins_df = origins.to_frame(name='origin')
```

```python
import duckdb
import time
```

```python
result = duckdb.sql("""
    SELECT origin FROM origins_df
""").fetchdf()
result
```

```python
start = time.time()
result = duckdb.sql("""
    SELECT *
    FROM read_csv_auto('../results/modified_files.csv', strict_mode=false, max_line_size=10000000, ignore_errors=true)
    WHERE origin IN (SELECT origin FROM origins_df)
""").fetchdf()
print("exec time: ", time.time() - start)
```

```python
result.to_csv("/home/infres/rapaport/git-historian/dataset-teaser/teaser_modified_files.csv", sep=';', index=False)
```

```python
with open("../results/teaser_modified_files.pkl", 'wb') as f:
    pickle.dump(result, f)
```
