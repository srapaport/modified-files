```python
import pandas as pd
import pickle
from gen_tex import update_latex_macros
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
```

```python
df = pd.read_csv("~/datasets/github-stars/github-stars.txt", header=None, names=['origin', 'stars'])
```

```python
with open('/home/infres/rapaport/datasets/github-stars/github-stars.pkl', 'wb') as f:
    pickle.dump(df, f)
```

```python
ds = pd.DataFrame
with open('/home/infres/rapaport/datasets/github-stars/github-stars.pkl', 'rb') as f:
    ds = pickle.load(f)
```

```python
ds.head()
```

```python
print(f'amount of entry for github stars: {len(ds):,}')
update_latex_macros("numbers.tex", "AmountStarsEntry", f'{len(ds):,}', "number")
```

```python
ds = ds.drop_duplicates().reset_index()
```

```python
print(f'amount of entry for github stars without duplicates: {len(ds):,}')
update_latex_macros("numbers.tex", "AmountStarsEntryWithoutDuplicates", f'{len(ds):,}', "number")
```

```python
ds['origin'] = ds['origin'].apply(lambda x: x.strip('"'))
```

```python
print(f'amount of origins: {len(ds):,}')
```

```python
star_0 = ds.loc[ds['stars'] == 0]
```

```python
print(f'amount of origins with 0 star: {len(star_0):,}')
update_latex_macros("numbers.tex", "AmountOri0Star", f'{len(star_0):,}', "number")
```

```python
star_1 = ds.loc[ds['stars'] == 1]
print(f'amount of origins with 1 star: {len(star_1):,}')
update_latex_macros("numbers.tex", "AmountOri1Star", f'{len(star_1):,}', "number")
```

```python
star_inco = ds.loc[ds['stars'] < 0]
print(f'amount of origins with less than 0 star: {len(star_inco):,}')
update_latex_macros("numbers.tex", "AmountOriInfStar", f'{len(star_inco):,}', "number")
```

```python
star_more1 = ds.loc[ds['stars'] > 1]
print(f'amount of origins with more than 1 star: {len(star_more1):,}')
update_latex_macros("numbers.tex", "AmountOriMoreThan1Star", f'{len(star_more1):,}', "number")
```

```python
with open('../results/stars_without_dup.pkl', 'wb') as f:
    pickle.dump(ds, f)
```
