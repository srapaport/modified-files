```python
result_grade = pd.DataFrame()
with open('../results_2026_03_02/grades.pkl', 'rb') as f:
    result_grade = pickle.load(f)
print(f'size of the dataframe: {len(result_grade):,}')
```

```python
result_grade.sample(5)
```

```python
new_grade = result_grade.loc[result_grade['amount_contrib']>1]
```

```python
# Find all rows in df where the origin exists in new_grade
matching_repos = pd.merge(df, result_grade, on='origin', how='inner')

# Display the result
print(f"Found {len(matching_repos)} repositories that exist in both dataframes")
matching_repos.sample(5)
```

```python
repos_with_rel = matching_repos.loc[matching_repos['amount_rel'] > 0]
repos_with_rel.sample(5)
```

```python
repos_modified = matching_repos.loc[matching_repos['status']=="Modified"]
len(repos_modified)
```

```python
def swhid_suffix(swhid: str, prefix: str) -> str:
    return swhid.split(f"{prefix}:", 1)[-1]

def repo_relative_path(path: str) -> str:
    return path[2:] if path.startswith("./") else path.lstrip("/")

def build_swh_license_urls(row) -> pd.Series:
    rev = swhid_suffix(row["revision"], "swh:1:rev")
    snap = swhid_suffix(row["snapshot_without"], "swh:1:snp")
    rel_path = repo_relative_path(row["path"])
    return pd.Series(
        {
            "Rev-License-Path": (
                f"https://archive.softwareheritage.org/browse/revision/{rev}/?path={rel_path}"
            ),
            "Snap-License-Path": (
                "https://archive.softwareheritage.org/browse/snapshot/"
                f"{snap}/directory/?branch={row['branch']}&path={rel_path}"
            ),
        }
    )

license_urls = repos_modified.apply(build_swh_license_urls, axis=1)
repos_modified_path = pd.concat([repos_modified, license_urls], axis=1)
repos_modified_path.to_csv("../results_2026_03_02/repos_modified_path.csv", sep=";", index=False)
repos_modified_path.to_csv("../results_2026_03_02/repos_modified_full.csv", sep=";", index=False)
print(f"Saved {len(repos_modified_path):,} rows with SWH browse URLs")
```

```python
repos_modified.to_csv('../results_2026_03_02/repos_modified.csv', index=False)

try:
    repos_modified.to_excel('../results_2026_03_02/repos_modified.xlsx', index=False, header=True)
    print(f"Saved {len(repos_modified)} to CSV and Excel files")
except Exception as e:
    print(f"Saved {len(repos_modified)} to CSV file only")
    print(f"Excel save failed: {e}")
```

```python
origins_impacted = pd.read_csv("../results_2026_03_02/origin_impacted.csv")
```

```python
origins_impacted.sample(5)
```

```python
rev_counts = origins_impacted['rev'].value_counts().reset_index()
rev_counts.columns = ['rev', 'count']

repos_with_impact_count = repos_with_rel.merge(
    rev_counts, 
    left_on='revision', 
    right_on='rev', 
    how='left'
)
repos_with_impact_count['count'] = repos_with_impact_count['count'].fillna(0).astype(int)
repos_with_impact_count = repos_with_impact_count.drop(columns=['rev'])

print(f"Total repos with releases: {len(repos_with_rel)}")
print(f"Repos with impact count > 1: {len(repos_with_impact_count[repos_with_impact_count['count'] > 1])}")
print(f"Maximum impact count: {repos_with_impact_count['count'].max()}")

repos_with_impact_count.sample(5)
```
